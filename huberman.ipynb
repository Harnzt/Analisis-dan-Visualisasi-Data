```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('haberman.csv')
```

```python
df.columns = ['Age', 'Year', 'Nodes', 'Survival']
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
sns.countplot(x='Survival', data=df)
plt.title('Distribusi Survival')
plt.xticks([0,1], ['Survived 5+ years', 'Died within 5 years'])
plt.show()
```

Distribusi Setiap Fitur

```python
sns.histplot(df['Age'], kde=True, bins=20)
plt.title('Distribusi Umur')
plt.show()
```

```python
sns.histplot(df['Nodes'], kde=True, bins=20)
plt.title('Distribusi Lymph Nodes')
plt.show()
```

```python
sns.histplot(df['Year'], kde=True, bins=20)
plt.title('Distribusi Tahun Operasi')
plt.show()
```

Analisis Survival vs Fitur

```python
sns.boxplot(x='Survival', y='Age', data=df)
plt.title('Age vs Survival')
plt.show()
```

```python
sns.boxplot(x='Survival', y='Nodes', data=df)
plt.title('Nodes vs Survival')
plt.show()
```

```python
sns.boxplot(x='Survival', y='Year', data=df)
plt.title('Year vs Survival')
plt.show()
```

```python
sns.pairplot(df, hue='Survival')
plt.show()
```

```python
corr = df.corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```
